#### 2023년 2월말 인구 데이터 _ 65세 인구현황

```python
up65_people0 = pd.read_excel('../P1_Data/01.External/2023년 2월말 인구.xlsx',sheet_name='65세이상인구현황',header=5,nrows=4)
up65_people = up65_people0[['공주시','유구읍','이인면','탄천면','계룡면','반포면','의당면','정안면','우성면','사곡면','신풍면','중학동','웅진동','금학동','옥룡동','신관동','월송동']]

```

```python
up65_people0
```

```python
up65_people = up65_people.drop(0)
up65_people.reset_index(drop=True)
df_transposed = up65_people.transpose()
df_transposed.columns = ['65세이상_합계','65세이상_남','65세이상_여']
# 인덱스를 컬럼으로 변경
df_transposed = df_transposed.reset_index()

# 인덱스 이름을 '지역명'으로 변경
df_transposed = df_transposed.rename(columns={'index': '행정기관'})
# 문자형을 정수형으로 변환
df_transposed[['65세이상_합계','65세이상_남','65세이상_여']] = df_transposed[['65세이상_합계','65세이상_남','65세이상_여']].replace(',', '', regex=True).astype(int)
df_transposed
#저장
#df_transposed.to_excel('../P1_Data/02.Pretreatment/pre_up65_pop.xlsx', index=False)
```
