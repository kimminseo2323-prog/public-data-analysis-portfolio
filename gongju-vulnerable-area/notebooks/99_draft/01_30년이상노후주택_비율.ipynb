#### 농촌지역 30년이상 노후주택 비율 _empty_house_sigol

```python
empty_house_sigol_old = empty_house_sigol[empty_house_sigol['건축년도'] <= 1993.0] #30년이상 노후주택 선택
empty_house_sigol_old
```

```python
empty_house_sigol_old_groupby = empty_house_sigol_old.groupby('행정동').agg(노후주택수=('빈집코드', 'count'))
empty_house_sigol_old_groupby = empty_house_sigol_old_groupby.reset_index()
empty_house_sigol_old_groupby = empty_house_sigol_old_groupby.rename(columns={'행정동': '행정기관'})
empty_house_sigol_old_groupby
```

```python
mapping = pd.merge(family_cnt,empty_house_sigol_old_groupby, on='행정기관')
mapping
```

```python
use_col = mapping[['행정기관','세대수', '노후주택수']]

# 비율 계산
df_ratio = use_col.apply(lambda x: (x['노후주택수'] / x['세대수']) * 100, axis=1)
df_ratio = df_ratio.apply(lambda x: round(x, 2))

# 새로운 데이터프레임 생성
result = pd.concat([use_col, df_ratio], axis=1)
result.columns = ['행정기관', '세대수','노후주택수', '노후주택 비율']

result.to_excel('../P3_Result/01.CSV/result_oldhouse_sigol_ratio.xlsx', index=False)
result
```

```python
df = result.copy()
ax = df.plot(x='행정기관',y=['세대수','노후주택수'],kind='bar',figsize=(10,6),title='공주시 농촌지역 30년이상 노후주택 비율',ylabel= '세대수',width=0.7)
plt.xticks(rotation=45)

# 특정컬럼 값들을 가져와서 막대 위에 표시
for i, row in df.iterrows():
    v = row['노후주택 비율']
    a = row['노후주택수']
    ax.text(i, a+100, f'{v:.2f}%', color='black', fontsize=10,ha='center')

plt.show()
```

```python
df = result.copy()
ax = df.plot(x='행정기관',y=['노후주택수'],kind='bar',figsize=(10,6),title='공주시 농촌지역 30년이상 노후주택 비율',ylabel= '노후주택수',width=0.7,color='orange')
plt.xticks(rotation=45)

# y축 범위 지정
plt.ylim(0, 80)

# 특정컬럼 값들을 가져와서 막대 위에 표시
for i, row in df.iterrows():
    v = row['노후주택 비율']
    a = row['노후주택수']
    ax.text(i, a+0.5, f'{v:.2f}%', color='black', fontsize=10,ha='center')

plt.show()
```

```python
# 파이차트로그리기
df_pie =result.copy()
df_pie.set_index('행정기관', inplace=True)
# 파이차트 그리기
df_pie['노후주택 비율'].plot(kind='pie', autopct='%1.1f%%', startangle=90, counterclock=False)
plt.axis('equal')
plt.legend(labels=df_pie.index, loc='upper right',bbox_to_anchor=(1.2, 0.5))
plt.title('공주시 농촌지역(읍,면) 노후주택 비율(30년이상 주택)')
plt.show()
```
