```python
def create_wordcloud(df, text_col, freq_col, figsize=(8, 8), bg_color='white', font_path='C:\\Windows\\Fonts\\NanumGothic.ttf'):
   
    # WordCloud 객체 생성
    wordcloud = WordCloud(width=800, height=800, background_color=bg_color, font_path=font_path).generate_from_frequencies(df.set_index(text_col)[freq_col].to_dict())

    # 워드클라우드 출력
    plt.figure(figsize=figsize)
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis('off')
    plt.show()
    
    return plt
```

```python
assis_top_30 = assis24_df.sort_values(by="빈도수", ascending=False).iloc[:30] #"빈도수" 열을 기준으로 내림차순 정렬 후 상위 30개 추출
assis_top_30['빈도수'] = assis_top_30['빈도수'] + np.mean(assis_top_30['빈도수'])*2 # 해당 열의 평균값의 2배를 더한 값으로 열을 대체 (값이 너무 적은경우를 대비하여 가중치 계산)
assis_top_30
```

```python
# create_wordcloud 함수 호출
asiis24_df_wordcloud = create_wordcloud(assis_top_30, 'new_extracted_word', '빈도수')
asiis24_df_wordcloud
# 워드클라우드 이미지 저장
asiis24_df_wordcloud.savefig('../P3_Result/02.PNG/3단계_보조금24_상위30개.png')
```
