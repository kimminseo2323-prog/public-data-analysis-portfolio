### 월별 각 기종 사용횟수 및 비율확인

```python
# 날짜 열을 datetime 형식으로 변환합니다
data['임대일자'] = pd.to_datetime(data['임대일자'])
# 날짜 열에서 월 정보를 추출하여 새로운 열을 생성합니다
data['월'] = data['임대일자'].dt.month
# 월별 각 기계의 사용 횟수를 카운트합니다
machine_usage_counts = data.groupby(['월', '기종']).size().reset_index(name='사용횟수')
# 월별 전체 임대건수를 계산합니다
total_monthly_counts = data.groupby('월').size().reset_index(name='전체건수')
# 각 기계의 월별 사용 비율을 계산합니다
machine_usage_ratio = pd.merge(machine_usage_counts, total_monthly_counts, on='월')
machine_usage_ratio['비율'] = machine_usage_ratio['사용횟수'] / machine_usage_ratio['전체건수']
machine_usage_ratio
```

```python
def plot_top_n_machines(data, month, n=5):
    # 월별로 필터링
    monthly_data = data[data['월'] == month]

    # 월별로 가장 많이 사용된 상위 n개 기종을 추출합니다
    top_n_machines = monthly_data.sort_values('사용횟수', ascending=False).head(n)

    # 막대 그래프를 그립니다
    plt.figure(figsize=(10, 6))
    bars = plt.bar(top_n_machines['기종'], top_n_machines['사용횟수'], color='lightblue')
    plt.xlabel('기종')
    plt.ylabel('사용횟수')
    plt.title(f'{month}월 상위 {n}개 기종 사용횟수')
    plt.xticks(rotation=0)

    # 범례 추가
    plt.legend(bars, [f'{machine} : {count}회' for machine, count in zip(top_n_machines['기종'], top_n_machines['사용횟수'])])

    plt.tight_layout()

    plt.show()
```

```python
# 월정보를 매개변수로 전달하여 함수 호출하여 막대 그래프 그리기
target_month = 1  # 원하는 월 정보 입력
plot_top_n_machines(machine_usage_ratio, month=target_month, n=5)
```

```python
# 월별 기계 사용 비율 중 가장 높은 비율을 기준으로 기계를 선택합니다
highest_ratio_per_month = machine_usage_ratio.groupby('월').apply(lambda x: x[x['비율'] == x['비율'].max()])
# 낮은 비율을 제거하여 선택한 기계들을 출력합니다
final_machine_selection = highest_ratio_per_month[highest_ratio_per_month['비율'] >= 0.1]
# 각 월별로 선택된 기계들을 출력합니다
# 결과를 데이터프레임으로 출력합니다
result_df = pd.DataFrame(columns=['월', '기종', '비율'])
for month, row in final_machine_selection.iterrows():
    result_df = result_df.append({'월': month, '기종': row['기종'], '비율': row['비율']}, ignore_index=True)
result_df
```

```python
# 날짜를 datetime 타입으로 변환
data['임대일자'] = pd.to_datetime(data['임대일자'])

# 기종별 월별 임대횟수 계산
data['월'] = data['임대일자'].dt.month
result_count = data.groupby(['기종', '월'])['임대금액'].count()

# 결과 출력을 위해 pivot_table로 재구성
result_df = result_count.reset_index().pivot_table(index='기종', columns='월', values='임대금액', fill_value=0)

# 컬럼명 변경 (1월부터 12월까지)
result_df.columns = [f'{month}월' for month in range(1, 13)]

result_df.to_csv('../P3_Result/01.CSV/시기별/시기별(월별)_기종임대횟수.csv',encoding='cp949',index = True)
result_df
```

```python
# 월별 데이터를 봄, 여름, 가을, 겨울로 묶기
seasons = {
    "봄": ["3월", "4월", "5월"],
    "여름": ["6월", "7월", "8월"],
    "가을": ["9월", "10월", "11월"],
    "겨울": ["12월", "1월", "2월"]
}

df_seasons = pd.DataFrame()
for season, months in seasons.items():
    df_seasons[season] = result_df[months].sum(axis=1)

df_seasons.to_csv('../P3_Result/01.CSV/시기별/시기별(계절)_기종임대횟수.csv',encoding='cp949',index = True)
df_seasons
```

```python
# 2.지역별 임대 현황 분석
data['진도군'] = data['진도군'].str.strip()
data['읍면'] = data['읍면'].str.strip()
data['마을'] = data['마을'].str.strip()
data['지역'] = data['진도군'] + ' ' + data['읍면'] + ' ' + data['마을']

# 기종별 임대 현황 분석
grouped_by_region = data.groupby('지역')
region_counts = grouped_by_region.size()
region_sum_duration = data.groupby('지역')['임대일수'].sum()
region_avg_duration = grouped_by_region['임대일수'].mean().round(1)
region_avg_amount = grouped_by_region['임대금액'].mean().round(1)

# 지역별로 기종별 임대 건수 계산
grouped_by_region_model = data.groupby(['지역', '기종'])
region_model_counts = grouped_by_region_model.size()

# 지역별로 임대가 가장 많은 기종 추출
top_models_by_region = region_model_counts.groupby('지역').idxmax().apply(lambda x: x[1])

# 결과를 하나의 데이터프레임으로 합치기
region_lease_status = pd.DataFrame({
    '임대건수': region_counts,
    '총 임대일수': region_sum_duration,
    '평균 임대일수': region_avg_duration,
    '평균 임대금액': region_avg_amount,
    '가장 많은 임대기종' : top_models_by_region
})
region_lease_status
```
