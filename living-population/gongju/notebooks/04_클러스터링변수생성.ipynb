```python
#클러스터링 데이터프레임 불러오기
ClusteredData = pd.read_excel('../P1_Data/02.Pretreatment/ClusteredData.xlsx')
ClusteredData
```

```python
ClusteredData_re = ClusteredData.copy()
```

```python
#2022년 12월 기준 주민등록인구
admin = pd.read_csv('../P1_Data/01.External/kosis/행정구역_읍면동_성별_5세별_주민등록인구_2022년12월기준.csv',encoding='cp949',skiprows=1)
admin = admin.rename(columns={'행정구역(동읍면)별':'행정동명','계':'주민등록인구'})
admin = admin[admin['항목'] == '총인구수 (명)']

admin['60세이상'] = admin[['60 - 64세','65 - 69세','70 - 74세','75 - 79세','80 - 84세','85 - 89세','90 - 94세','95 - 99세','100+']].sum(axis=1)
admin['주민등록인구_60세이상_비율'] = ((admin['60세이상'] / admin['주민등록인구']) * 100).round(1)
admin = admin[['행정동명','주민등록인구','주민등록인구_60세이상_비율']]

flowpop = pd.read_csv('../P3_Result/01.CSV/2.행정동별_생활인구패턴/행정동별_성별유동인구(2022.12월기준).csv',encoding='cp949')
flowpop = flowpop[['ADM_NM','성연령별_유동인구']]
flowpop.columns = ['행정동명','유동인구']
flowpop2 = pd.merge(flowpop,admin,on='행정동명')
flowpop2['유동인구_배율'] = (flowpop2['유동인구'] / flowpop2['주민등록인구']).round(1)
```

```python
ClusteredData_re = pd.merge(ClusteredData_re,flowpop2,on = '행정동명')
ClusteredData_re
```

```python
flowpop_age = pd.read_csv('../P3_Result/01.CSV/2.행정동별_생활인구패턴/행정동별_연령대별유동인구(2022.12월기준).csv',encoding='cp949')
flowpop_age = flowpop_age[['ADM_NM','60대이상_비율','20대_비율']]
flowpop_age = flowpop_age.rename(columns={'ADM_NM':'행정동명','60대이상_비율':'유동인구_60대이상_비율','20대_비율':'유동인구_20대_비율'})
flowpop_age
```

```python
ClusteredData_re = pd.merge(ClusteredData_re,flowpop_age,on = '행정동명')
ClusteredData_re
```

```python
GrowthRate = pd.read_csv('../P3_Result/01.CSV/2.행정동별_생활인구패턴/행정동별_월평균유동인구_증가율.csv',encoding='cp949')
GrowthRate = GrowthRate[['행정동명','증가율(%)']]
GrowthRate = GrowthRate.rename(columns={'증가율(%)':'유동인구_증가율(%)'})
GrowthRate
```

```python
ClusteredData_re = pd.merge(ClusteredData_re,GrowthRate,on = '행정동명')
ClusteredData_re
```

```python
Concentration_day = pd.read_csv('../P3_Result/01.CSV/3.집중도/행정동별_주중주말집중도.csv',encoding='cp949')
Concentration_day = Concentration_day[['ADM_NM','주중_집중도']]
Concentration_day = Concentration_day.rename(columns={'ADM_NM':'행정동명'})

Concentration_time = pd.read_csv('../P3_Result/01.CSV/3.집중도/행정동별_특정시간집중도.csv',encoding='cp949')
Concentration_time = Concentration_time[['ADM_NM','오후12시_집중도']]
Concentration_time = Concentration_time.rename(columns={'ADM_NM':'행정동명'})
Concentration = pd.merge(Concentration_day,Concentration_time,on='행정동명')
Concentration
```

```python
ClusteredData_re = pd.merge(ClusteredData_re,Concentration,on = '행정동명')
ClusteredData_re
```

```python
ClusteredData_re.to_csv('../P1_Data/02.Pretreatment/ClusteredData_result.csv',encoding='cp949',index=False)
```
