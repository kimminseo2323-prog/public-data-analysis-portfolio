### 행정동별 연령대별 유동인구

```python
HJD_age_pop = HJD_ax_age_pop[HJD_ax_age_pop['CRTR_YM'] == '202212']
HJD_age_pop = HJD_age_pop.groupby(['CRTR_YM', 'YEAR', 'MONTH', 'ADM_NM']).sum().round(1).reset_index()

HJD_age_pop['M00_19'] = HJD_age_pop[['ML_FLOW_PPLTN_CNT_10_BELOW','ML_FLOW_PPLTN_CNT_10_14', 'ML_FLOW_PPLTN_CNT_15_19']].sum(axis=1)
HJD_age_pop['M20_29'] = HJD_age_pop[['ML_FLOW_PPLTN_CNT_20_24', 'ML_FLOW_PPLTN_CNT_25_29']].sum(axis=1)
HJD_age_pop['M30_39'] = HJD_age_pop[['ML_FLOW_PPLTN_CNT_30_34', 'ML_FLOW_PPLTN_CNT_35_39']].sum(axis=1)
HJD_age_pop['M40_49'] = HJD_age_pop[['ML_FLOW_PPLTN_CNT_40_44', 'ML_FLOW_PPLTN_CNT_45_49']].sum(axis=1)
HJD_age_pop['M50_59'] = HJD_age_pop[['ML_FLOW_PPLTN_CNT_50_54', 'ML_FLOW_PPLTN_CNT_55_59']].sum(axis=1)
HJD_age_pop['M60_UP'] = HJD_age_pop[['ML_FLOW_PPLTN_CNT_60_64', 'ML_FLOW_PPLTN_CNT_65_69','ML_FLOW_PPLTN_CNT_70_ABV']].sum(axis=1)

HJD_age_pop['F00_19'] = HJD_age_pop[['FM_FLOW_PPLTN_CNT_10_BELOW','FM_FLOW_PPLTN_CNT_10_14', 'FM_FLOW_PPLTN_CNT_10_14']].sum(axis=1)
HJD_age_pop['F20_29'] = HJD_age_pop[['FM_FLOW_PPLTN_CNT_20_24', 'FM_FLOW_PPLTN_CNT_25_29']].sum(axis=1)
HJD_age_pop['F30_39'] = HJD_age_pop[['FM_FLOW_PPLTN_CNT_30_34', 'FM_FLOW_PPLTN_CNT_35_39']].sum(axis=1)
HJD_age_pop['F40_49'] = HJD_age_pop[['FM_FLOW_PPLTN_CNT_40_44', 'FM_FLOW_PPLTN_CNT_45_49']].sum(axis=1)
HJD_age_pop['F50_59'] = HJD_age_pop[['FM_FLOW_PPLTN_CNT_50_54', 'FM_FLOW_PPLTN_CNT_55_59']].sum(axis=1)
HJD_age_pop['F60_UP'] = HJD_age_pop[['FM_FLOW_PPLTN_CNT_60_64', 'FM_FLOW_PPLTN_CNT_65_69','FM_FLOW_PPLTN_CNT_70_ABV']].sum(axis=1)

HJD_age_pop['T00_19'] = HJD_age_pop['M00_19'] + HJD_age_pop['F00_19']
HJD_age_pop['T20_29'] = HJD_age_pop['M20_29'] + HJD_age_pop['F20_29']
HJD_age_pop['T30_39'] = HJD_age_pop['M30_39'] + HJD_age_pop['F30_39']
HJD_age_pop['T40_49'] = HJD_age_pop['M40_49'] + HJD_age_pop['F40_49']
HJD_age_pop['T50_59'] = HJD_age_pop['M50_59'] + HJD_age_pop['F50_59']
HJD_age_pop['T60_up'] = HJD_age_pop['M60_UP'] + HJD_age_pop['F60_UP']
HJD_age_pop = HJD_age_pop[['CRTR_YM', 'YEAR', 'MONTH', 'ADM_NM', 'T00_19', 'T20_29', 'T30_39', 'T40_49','T50_59', 'T60_up']]
HJD_age_pop['유동인구'] = HJD_age_pop[['T00_19', 'T20_29', 'T30_39', 'T40_49','T50_59', 'T60_up']].sum(axis=1)

HJD_age_pop['10대_비율'] = (HJD_age_pop['T00_19'] / HJD_age_pop['유동인구'] * 100).round(1)
HJD_age_pop['20대_비율'] = (HJD_age_pop['T20_29'] / HJD_age_pop['유동인구'] * 100).round(1)
HJD_age_pop['30대_비율'] = (HJD_age_pop['T30_39'] / HJD_age_pop['유동인구'] * 100).round(1)
HJD_age_pop['40대_비율'] = (HJD_age_pop['T40_49'] / HJD_age_pop['유동인구'] * 100).round(1)
HJD_age_pop['50대_비율'] = (HJD_age_pop['T50_59'] / HJD_age_pop['유동인구'] * 100).round(1)
HJD_age_pop['60대이상_비율'] = (HJD_age_pop['T60_up'] / HJD_age_pop['유동인구'] * 100).round(1)
```

```python
mapping = gpd.sjoin(pre_result, gongju_boundary,how='left')

# 특정 컬럼 값이 중복되는 행 추출
duplicate_rows = mapping[mapping.duplicated(subset='gid', keep=False)]
#mapping한 기존격자에서 중복 되는 행 제거
mapping = mapping.drop_duplicates(subset='gid', keep=False)


##중복되는 행의 처리를 위한 작업
#중복되는 행들과 시흥 경계면 오버레이하여 걸쳐진 부분 면적 구함
res_union = gpd.overlay(duplicate_rows, gongju_boundary, how='intersection')

# 두 컬럼의 값이 동일한 행 추출
same_res_union = res_union[res_union['ADM_NM_1'] == res_union['ADM_NM_2']]

#면적계산
same_res_union['area'] = same_res_union['geometry'].area

# 같은 GID 값끼리 그룹화하여 면적컬럼에서 가장 큰 값을 가진 읍면동컬럼으로 업데이트
result_union = same_res_union.groupby('gid').apply(lambda x: x.loc[x['area'].idxmax()]).reset_index(drop=True)
# 행정동 매핑한 격자와 여러개의 행정동이 걸처진 격자처리 후 두개의 데이터프레임을 다시 합치는 과정
mapping = mapping[['gid','ADM_CD','ADM_NM','주중','주말','geometry']]

result_union = result_union[['gid', 'ADM_CD_1', 'ADM_NM_1', '주중','주말', 'geometry']]
# 'A' 컬럼의 이름을 'X'로 변경
result_union.rename(columns={'ADM_CD_1': 'ADM_CD'}, inplace=True)
result_union.rename(columns={'ADM_NM_1': 'ADM_NM'}, inplace=True)

#결과
focus_day_result = pd.concat([mapping, result_union], axis=0)
# 인덱스를 리셋합니다.
focus_day_result = focus_day_result.reset_index(drop=True)
```

```python
# 집중도
focus_day_result['주중_집중도'] = focus_day_result['주중'] / (focus_day_result['주중'] + focus_day_result['주말'])
focus_day_result['주중_집중도'] = focus_day_result['주중_집중도'].fillna(0)  # NaN 값을 0으로 대체합니다.

focus_day_result['주말_집중도'] = focus_day_result['주말'] / (focus_day_result['주말'] + focus_day_result['주중'])
focus_day_result['주말_집중도'] = focus_day_result['주말_집중도'].fillna(0)  # NaN 값을 0으로 대체합니다.
focus_day_result.insert(5, '주중_집중도', focus_day_result.pop('주중_집중도'))
focus_day_result.insert(6, '주말_집중도', focus_day_result.pop('주말_집중도'))
# 격자별 집중도 파일 저장
focus_day_result.to_file('../P3_Result/03.SHP/공주시_주중주말집중도/공주시_주중주말_격자별집중도.gpkg')

# 그룹화된 결과에 주중과 주말 컬럼의 합계 계산
focus_day = focus_day_result.groupby(['ADM_NM'])[['주중_집중도', '주말_집중도', '주중', '주말']].agg({'주중': 'sum', '주말': 'sum', '주중_집중도': 'mean', '주말_집중도': 'mean'}).round(2)
focus_day.to_csv('../P3_Result/01.CSV/3.집중도/행정동별_주중주말집중도.csv', encoding = 'cp949')
```
