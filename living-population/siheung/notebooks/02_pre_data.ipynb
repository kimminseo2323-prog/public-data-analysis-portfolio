```python
import pandas as pd
import geopandas as gpd
```

#### 시흥시 행정경계 데이터

```python
siheung = gpd.read_file('../P1_Data/02.Pretreatment/siheung_boundary.shp', encoding='cp949')
siheung
```

#### 격자에 행정동 매핑한 데이터

```python
siheung_grid_2020 = gpd.read_file('../P3_Result/03.SHP/시흥시_100M격자_생활인구(행정동)_2020년도.shp', encoding = 'cp949')
siheung_grid_2021 = gpd.read_file('../P3_Result/03.SHP/시흥시_100M격자_생활인구(행정동)_2021년도.shp', encoding = 'cp949')
siheung_grid_2022 = gpd.read_file('../P3_Result/03.SHP/시흥시_100M격자_생활인구(행정동)_2022년도.shp', encoding = 'cp949')
siheung_grid_2023 = gpd.read_file('../P3_Result/03.SHP/시흥시_100M격자_생활인구(행정동)_2023년도.shp', encoding = 'cp949')
siheung_grid = siheung_grid_2020[['GRID_100M_','ADM_CD','ADM_NM']]
siheung_grid
```

### 생활인구 데이터


```python
# 성연령별 2020 ~ 2023
merged_sex_age = gpd.read_file('../P1_Data/02.Pretreatment/merged_sex_age(20~23).shp',encoding='cp949')
# 시간별 2020 ~ 2023
merged_time = gpd.read_file('../P1_Data/02.Pretreatment/merged_time(20~23).shp',encoding='cp949')
# 요일별 2020 ~ 2023
merged_day = gpd.read_file('../P1_Data/02.Pretreatment/merged_day(20~23).shp',encoding='cp949')
```

```python
#복사
pop_gender_age = merged_sex_age.copy()
pop_time = merged_time.copy()
pop_day = merged_day.copy()
```

```python
#성연령별 행정동 매핑
siheung_grid_gender = pd.merge(siheung_grid,pop_gender_age,on='GRID_100M_')
siheung_grid_time = pd.merge(siheung_grid,pop_time,on='GRID_100M_')
siheung_grid_day = pd.merge(siheung_grid,pop_day,on='GRID_100M_')
```

```python
siheung_grid_gender = gpd.GeoDataFrame(siheung_grid_gender,geometry='geometry', crs='EPSG:5179')
siheung_grid_time = gpd.GeoDataFrame(siheung_grid_time,geometry='geometry', crs='EPSG:5179')
siheung_grid_day = gpd.GeoDataFrame(siheung_grid_day,geometry='geometry', crs='EPSG:5179')
```

```python
siheung_grid_gender.to_file('../P3_Result/03.SHP/시흥시_100M격자_성연령별_생활인구(행정동).shp', encoding = 'cp949')
siheung_grid_time.to_file('../P3_Result/03.SHP/시흥시_100M격자_시간대별_생활인구(행정동).shp', encoding = 'cp949')
siheung_grid_day.to_file('../P3_Result/03.SHP/시흥시_100M격자_요일별_생활인구(행정동).shp', encoding = 'cp949')
```
